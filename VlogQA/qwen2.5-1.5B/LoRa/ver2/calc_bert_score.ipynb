# Tính BERTScore cho kết quả dự đoán
Notebook này sẽ đọc kết quả từ file JSON và tính toán độ đo BERTScore bằng thư viện `bert_score`.

```python
import json
from bert_score import score

# Đường dẫn tới file JSON
file_path = r"c:\Users\Admin\Desktop\Long\repos\PACLIC_2026\VlogQA\qwen2.5-3B\LoRa\Ver2\eval_results_vlogqa_3b_2.json"

with open(file_path, 'r', encoding='utf-8') as f:
    data = json.load(f)

preds = []
refs = []

for item in data.get("predictions", []):
    # Sử dụng 'clean_prediction' để đánh giá so với 'ground_truth'
    preds.append(item.get("clean_prediction", ""))
    refs.append(item.get("ground_truth", ""))

print(f"Tổng số mẫu đánh giá: {len(preds)}")
```

```python
# Tính toán BERTScore với ngôn ngữ tiếng Việt (lang='vi')
P, R, F1 = score(preds, refs, lang='vi', verbose=True)

print(f"\nKết quả BERTScore Trung Bình:")
print(f"Precision: {P.mean().item():.4f}")
print(f"Recall:    {R.mean().item():.4f}")
print(f"F1-Score:  {F1.mean().item():.4f}")
```
